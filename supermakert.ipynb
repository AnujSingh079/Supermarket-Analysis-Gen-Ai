```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv(r"C:\Users\Anuj Singh\Downloads\supermarket.csv")
df
```

```python
#Question 1 Counting the Gender Count
g=df['Gender'].value_counts()
g
```

```python
#Question 3
df["Branch"].value_counts()
```

```python
#Q4 most popular payment method

sns.countplot(x=df['Payment'])
plt.title("count vs Method Of Payment")
plt.xlabel("Method Of Payment")
plt.ylabel("count")
```

```python
#Q4b Find Branch Wise Payment Method distribution in all branches

plt.figure(figsize=(14,6))
plt.style.use('classic')
ax = sns.countplot(x="Payment",hue="Branch",data=df,palette="tab20")
ax.set_title(label="Payment distribution in all branches",fontsize=20)
ax.set_xlabel(xlabel="payment Method",fontsize=16)
ax.set_ylabel(ylabel="People Count",fontsize=16)
```

```python
plt.figure(figsize=(14,6))
plt.style.use('classic')
ax = sns.countplot(x="Customer type",hue="Branch",data=df,palette="rocket_r")
ax.set_title(label="Customer Type in different branch",fontsize=25)
ax.set_xlabel(xlabel="Branches",fontsize=16)
ax.set_ylabel(ylabel="Customer Count",fontsize=16)
```

```python
#Q5 Does gross income affect customer ratings

sns.set_style('darkgrid')
sns.scatterplot(x=df['Rating'],y=df['gross income'])
plt.title("Gross Income Vs Rating")
```

```python
#Q6 Find Rating distribution between branches #Boxplot is used when we have to look b/w categorical and numberical data

plt.figure(figsize=(8,4))
ax = sns.boxplot(x="Branch",y="Rating",data=df,palette="RdYlBu")
ax.set_title("Rating distribution between branches",fontsize=25)
ax.set_xlabel(xlabel="Branches",fontsize=16)
ax.set_ylabel(ylabel="Rating distribution",fontsize=16)
plt.grid()
```

```python
#Q7 Which is the most profitabble branch?

sns.boxplot(x=df['Branch'],y=df['gross income'])
plt.title("Gross Income Vs Branch")
```

```python
#Q8 Relationship between Gender and Gross income

sns.boxplot(x=df['Gender'],y=df['gross income'])
plt.title("Gross income vs gender")
```

```python
#Q9 Can you see any time trend in gross income? 

df.groupby(df.index).mean(numeric_only=True)
```

```python
sns.lineplot(x=df.groupby(df.index).mean(numeric_only=True).index,
             y=df.groupby(df.index).mean(numeric_only=True)['gross income'])
```

```python
#Q10 Which product line helps supermarket generate the most income? 

cat = df[["Product line","gross income"]].groupby(['Product line'],as_index=False).sum().sort_values(by="gross income",ascending=False)
plt.figure(figsize=(20,8))
sns.barplot(x="Product line",y="gross income",data=cat)
plt.title("Gross income vs Product line")
```
